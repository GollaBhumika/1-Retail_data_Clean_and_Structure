###  Prepare Data for SQL Analysis

```python
df["order_date"] = pd.to_datetime(
    df["order_date"], errors="coerce"
)

# Check invalid dates
print("Missing or invalid order dates:", df["order_date"].isna().sum())

# Save dates in a SQLite-friendly format
df["order_date"] = df["order_date"].dt.strftime("%Y-%m-%d")

print("Date preparation completed!")
display(df.head())
```

# five SQL queries

# Query 1 — Top 10 Products by Sales
Business question: Which products generate the highest total sales?

```python
query1 = """
SELECT
    product_name,
    ROUND(SUM(sales), 2) AS total_sales
FROM sales
GROUP BY product_name
ORDER BY total_sales DESC
LIMIT 10;
"""

result1 = pd.read_sql(query1, conn)
display(result1)
```

# Query 2 — Monthly Sales Trend
Business question: How do sales change month by month?

```python
query2 = """
SELECT
    strftime('%Y-%m', order_date) AS sales_month,
    ROUND(SUM(sales), 2) AS total_sales
FROM sales
WHERE order_date IS NOT NULL
GROUP BY strftime('%Y-%m', order_date)
ORDER BY sales_month;
"""

result2 = pd.read_sql(query2, conn)
display(result2)
```

# Query 3 — Sales by Region
Business question: Which regions generate the highest sales?

```python
query3 = """
SELECT
    region,
    ROUND(SUM(sales), 2) AS total_sales
FROM sales
GROUP BY region
ORDER BY total_sales DESC;
"""

result3 = pd.read_sql(query3, conn)
display(result3)
```

# Query 4 — Average Order Value by Customer Segment
Business question: What is the average order value for each customer segment?

```python
query4 = """
SELECT
    segment,
    ROUND(AVG(order_value), 2) AS average_order_value
FROM (
    SELECT
        order_id,
        segment,
        SUM(sales) AS order_value
    FROM sales
    GROUP BY order_id, segment
)
GROUP BY segment
ORDER BY average_order_value DESC;
"""

result4 = pd.read_sql(query4, conn)
display(result4)
```

# Query 5 — Sales by Product Category
Business question: Which product categories contribute the most to sales?

```python
query5 = """
SELECT
    category,
    ROUND(SUM(sales), 2) AS total_sales,
    ROUND(AVG(sales), 2) AS average_sales
FROM sales
GROUP BY category
ORDER BY total_sales DESC;
"""

result5 = pd.read_sql(query5, conn)
display(result5)
```
